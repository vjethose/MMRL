```python
# 3c. ROCOv2 (latest Zenodo version; train/val drawn disjointly from the official valid split, test from the official test split)
import requests, os, zipfile, json
rec = requests.get('https://zenodo.org/api/records/8333645/versions/latest', timeout=60).json()
print('ROCOv2 record', rec.get('id'), rec.get('metadata', {}).get('version'), rec.get('metadata', {}).get('publication_date'))
files = {f['key']: f['links']['self'] for f in rec['files']}
os.makedirs('/tmp/roco', exist_ok=True)
for k in ['cui_mapping.csv', 'valid_captions.csv', 'valid_concepts.csv', 'test_captions.csv', 'test_concepts.csv', 'valid_images.zip', 'test_images.zip']:
    dst = f'/tmp/roco/{k}'
    if not os.path.exists(dst):
        with requests.get(files[k], stream=True, timeout=600) as r:
            r.raise_for_status()
            with open(dst, 'wb') as f:
                for chunk in r.iter_content(1 << 22): f.write(chunk)
    print(k, os.path.getsize(dst) // 2**20, 'MB')
    if k.endswith('.zip'):
        zipfile.ZipFile(dst).extractall(f'/tmp/roco/{k[:-4]}')
json.dump({'record': rec.get('id'), 'version': rec.get('metadata', {}).get('version')}, open('/tmp/roco/record.json', 'w'))
!python tools/prepare_rocov2.py --root /tmp/roco --out /kaggle/working/data/rocov2 --split_map train=valid,val=valid,test=test --max_train 5000 --max_eval 2000 --copy_images
!cp /tmp/roco/record.json /kaggle/working/data/rocov2/zenodo_record.json
```

```python
# 5. the grid: GPU 0 runs ScienceQA + ablations, GPU 1 runs ogbn-arxiv + ROCOv2
import subprocess, time
env = dict(os.environ, AMP='fp16', BS='16', ACCUM='2', EPOCHS='4', PATIENCE='2', SEEDS='0 1 2', WORKERS='2')
def job(gpu, stages, log):
    cmd = ' ; '.join(f'STAGE={s} bash kaggle/session.sh' for s in stages)
    return subprocess.Popen(f'CUDA_VISIBLE_DEVICES={gpu} bash -c "{cmd}" > {log} 2>&1', shell=True, env=env)
ngpu = torch.cuda.device_count()
if ngpu >= 2:
    jobs = [job(0, ['sqa', 'ablation'], '/kaggle/working/log_gpu0.txt'), job(1, ['arxiv', 'roco'], '/kaggle/working/log_gpu1.txt')]
else:
    jobs = [job(0, ['sqa', 'arxiv', 'roco', 'ablation'], '/kaggle/working/log_gpu0.txt')]
t0 = time.time()
DEADLINE_H = 10.5  # Kaggle stops a commit at 12 h; leave time for cost + report and keep finished runs
while any(j.poll() is None for j in jobs):
    time.sleep(600)
    if (time.time() - t0) / 3600 > DEADLINE_H:
        print('deadline reached: stopping remaining runs; finished runs are kept', flush=True)
        subprocess.run('pkill -f train.py', shell=True)
        for j in jobs: j.terminate()
        break
    n = len([f for f in os.listdir('/kaggle/working/results') if f.endswith('.json')]) if os.path.exists('/kaggle/working/results') else 0
    print(f'{(time.time()-t0)/3600:.1f} h, {n} runs finished', flush=True)
print('grid done in', round((time.time()-t0)/3600, 2), 'h')
!grep -h -E "FAILED|Error" /kaggle/working/log_gpu*.txt | tail -20
```
